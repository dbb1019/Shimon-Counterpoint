```python
import music21
import csv

# Load the score from a MusicXML file
score = music21.converter.parse('/Users/annie/Desktop/Chorales/midiCmaj/4_4/bwv52.6_transposed.mid')

# Get the measures in the score
measures = score.getElementsByClass(music21.stream.Part).elements[0].getElementsByClass(music21.stream.Measure)

# Slice the measures into 12-bar units
bars_per_slice = 12
slices = [measures[i:i+bars_per_slice] for i in range(0, len(measures), bars_per_slice)]

# Loop through each slice and save it to a separate CSV file
for i, slice in enumerate(slices):
    # Create a list to store the data
    data = []
    # Loop through each measure in the slice
    for measure in slice:
        # Get the measure number
        measure_number = measure.measureNumber
        # Get the measure line type at the start of the measure
        measure_line_start = None
        barlines = measure.getElementsByClass(music21.bar.Barline)
        if barlines:
            measure_line_start = barlines[0].type
            
        # Get the measure line type at the end of the measure
        measure_line_end = None
        if barlines:
            measure_line_end = barlines[-1].type

        # Loop through each note in the measure
        for note in measure.flat.notes:
            # Get the MIDI note number
            note_number = note.pitch.midi
            # Get the duration of the note in quarter notes
            duration = note.duration.quarterLength
            # Append the data to the list
            data.append([measure_number, measure_line_start, measure_line_end, note_number, duration])
        # Save the data to a CSV file
        filename = f'/Users/annie/Desktop/test{i}.csv'
        with open(filename, 'w', newline='') as f:
            writer = csv.writer(f)
            writer.writerow(['Measure', 'Measure Line Start', 'Measure Line End', 'Note Number', 'Duration'])
            writer.writerows(data)
```
