```python
import os
import csv
from music21 import *
```

```python
# extract key, tempo, time signature info from a midi file
def extract_info(midi_path):
    midi_file = converter.parse(midi_path)
    tempo = midi_file.recurse().getElementsByClass('MetronomeMark')[0].number
    key = midi_file.analyze('key')
    time_signature = midi_file.getTimeSignatures()[0]
    
    print(midi_path, key, tempo, time_signature)
    
    return (key, tempo, time_signature)
```

```python
# normalize midi file to C major and A minor
def midi_transpose(input_path, output_path):
    # load MIDI file or music score
    music_file = converter.parse(input_path)
    
    # get the key and mode of the input file
    key = music_file.analyze('key')
    mode = key.mode

    if mode == 'major':
        key_diff = interval.Interval(key.tonic, pitch.Pitch('C'))  # get the interval to C major
    elif mode == 'minor':
        key_diff = interval.Interval(key.tonic, pitch.Pitch('A'))  # get the interval to A minor
    else:
        raise ValueError('Unsupported mode: {}'.format(mode))
    #print(key_diff)
    music_file.transpose(key_diff, inPlace=True)  # transpose all notes in-place

    # save transposed score to a new MIDI file
    music_file.write('midi', output_path)
```

```python
info_list = []

for i in range(1, 16):
    file_path = f'/Users/annie/Desktop/Chorales/original/chorale{i}.mid'

    output_path = f'/Users/annie/Desktop/Chorales/transpose/chorale{i}.mid'

    midi_transpose(file_path, output_path)

    info = extract_info(file_path)
    info_list.append(info)
    csv_file = '/Users/annie/Desktop/Chorales/original/info.csv'

    with open(csv_file, 'w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['MIDI File', 'Key', 'Tempo', 'Time Signature'])
        for i, info in enumerate(info_list):
            writer.writerow([f'chorale{i+1}', info[0], info[1], info[2]])
```
