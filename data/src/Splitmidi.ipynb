```python
import os
import shutil
from music21 import converter, key, meter
```

```python
def split_major_minor_midi_files(input_dir, output_dir):
    # Create output directories if they don't exist
    major_dir = os.path.join(output_dir, "midiCmaj")
    minor_dir = os.path.join(output_dir, "midiamin")
    os.makedirs(major_dir, exist_ok=True)
    os.makedirs(minor_dir, exist_ok=True)

    # Iterate over MIDI files in input directory
    for file_name in os.listdir(input_dir):
        if file_name.endswith(".mid"):
            # Load MIDI file and determine key
            file_path = os.path.join(input_dir, file_name)
            midi_stream = converter.parse(file_path)
            midi_key = midi_stream.analyze("key")

            # Determine whether key is major or minor
            if midi_key.mode == "major":
                output_path = os.path.join(major_dir, file_name)
            else:
                output_path = os.path.join(minor_dir, file_name)

            # Copy file to appropriate output directory
            shutil.copy(file_path, output_path)
```

```python
input_dir = "/Users/annie/Desktop/Chorales/midi/"
output_dir = "/Users/annie/Desktop/Chorales/"
split_major_minor_midi_files(input_dir, output_dir)
```

```python
def split_4_4_midi_files(input_dir, output_dir):
    # Create output directories if they don't exist
    four_four_dir = os.path.join(output_dir, "4_4")
    other_dir = os.path.join(output_dir, "other")
    os.makedirs(four_four_dir, exist_ok=True)
    os.makedirs(other_dir, exist_ok=True)

    # Iterate over MIDI files in input directory
    for file_name in os.listdir(input_dir):
        if file_name.endswith(".mid"):
            # Load MIDI file and determine time signature
            file_path = os.path.join(input_dir, file_name)
            midi_stream = converter.parse(file_path)
            midi_time_signature = midi_stream.flat.getElementsByClass(meter.TimeSignature)[0]

            # Determine whether time signature is 4/4 or not
            if midi_time_signature.ratioString == "4/4":
                output_path = os.path.join(four_four_dir, file_name)
            else:
                output_path = os.path.join(other_dir, file_name)

            # Copy file to appropriate output directory
            shutil.copy(file_path, output_path)
```

```python
input_dir = "/Users/annie/Desktop/Chorales/midiamin/"
output_dir = "/Users/annie/Desktop/Chorales/midiamin/"
split_4_4_midi_files(input_dir, output_dir)
```
